## What you will find

The tutorials live in the [physicalai](https://github.com/openvinotoolkit/physicalai) repository. The reference notebook points to both published tutorial pages:

- **[Collect → Train → Deploy walkthrough](https://github.com/openvinotoolkit/physicalai/blob/main/examples/tutorials/README.md)** — a single end-to-end tutorial: collect teleoperation data, train a visuomotor policy, export it to OpenVINO, and deploy it on a real robot.
- **[Full tutorial notebook collection](https://github.com/openvinotoolkit/physicalai/blob/docs/tutorials/examples/tutorials/README.md)** — a broader set of notebooks: an introduction, using Physical AI Studio, OpenVINO optimization, testing deployment without a robot, and a full data-collection-to-deployment workflow.

Refer to the physicalai repository for the current list and the step-by-step instructions.

## Requirements

Running the tutorials needs Intel® hardware for OpenVINO plus robotics hardware (a robot arm and a camera). The **authoritative, up-to-date requirements and setup steps live in the [physicalai](https://github.com/openvinotoolkit/physicalai) repository** — please follow them there rather than a copy that can drift.

At a glance, you will need an OpenVINO-supported Intel® platform (CPU, GPU, or NPU) and a GPU for policy training; see the [OpenVINO System Requirements](https://docs.openvino.ai/2026/about-openvino/release-notes-openvino/system-requirements.html) and the physicalai docs for exact details.
