```python
import ipywidgets as widgets
import openvino as ov
from optimum.intel.openvino import OVLatentConsistencyModelPipeline
import gc

# Initialize OpenVINO Core
core = ov.Core()
```

```python
# Create a dropdown widget to select the device (CPU, GPU, AUTO)
device = widgets.Dropdown(
    options=core.available_devices + ["AUTO"],
    value='AUTO',
    description='Device:',
    disabled=False,
)

device
```

```python
model_id = "SimianLuo/LCM_Dreamshaper_v7"

print(f"Loading {model_id} and converting to OpenVINO IR... (This may take a few minutes)")

# Load the model and export it to OpenVINO format automatically
pipe = OVLatentConsistencyModelPipeline.from_pretrained(
    model_id,
    export=True,
    device=device.value 
)

# Compile the model explicitly for the selected device
pipe.compile()
print(f"✅ Model successfully compiled for {device.value}")
```
